```python
# Problema 1
for numero in range(-30, 51):
  if numero < 0:
     print(f"{numero} Muy frío")
  elif numero >= 0 and numero <= 24:
    print(f"{numero} Templado")
  else:
    print(f"{numero} Caluroso")
```

```python
# Problema 2
for numero in range(1, 101):
  if numero % 5 == 0:
    if numero % 10 == 0:
      print(f"{numero} Bip Bip")
    else:
      print(f"{numero} Bip")
```

```python
# Problema 3
puntaje_1 = int(input("Ingresa primer puntaje: "))
puntaje_2 = int(input("Ingresa segundo puntaje: "))
suma = puntaje_1 + puntaje_2

if suma < 50:
  print("/nResultado: Nivel principiante")
elif suma >= 50 and suma < 80:
  print("/nResultado: Nivel intermedio")
else:
  print("/nResultado: Nivel pro")
```

```python
# Problema 4
edad = int(input("Ingresa tu edad: "))
respuesta_musica = input("¿Te gusta la música? (si/no): ").lower()
respuesta_bailar = input("¿Te gusta bailar (si/no): ").lower()

le_gusta_musica = (respuesta_musica == "si")
le_gusta_bailar = (respuesta_bailar == "si")

if edad >= 18 and le_gusta_musica and le_gusta_bailar:
  print("/nEres una estrella de la fiesta")
elif le_gusta_musica and not le_gusta_bailar:
  print("/nte gusta la música, pero no tanto bailar")
else:
  print("nTienes tu propio estilo")
```
